```python
import requests
from bs4 import BeautifulSoup
import os
```

```python
ligues = {
    "ARGENTINA_A": [
        "https://www.transfermarkt.us/mamelodi-sundowns-fc/startseite/verein/6356",
        "https://www.transfermarkt.us/river-plate/startseite/verein/223",
    ],
    "ARGENTINA_B": [
        "https://www.transfermarkt.us/mamelodi-sundowns-fc/startseite/verein/6356",
        "https://www.transfermarkt.us/river-plate/startseite/verein/223",
    ],
}

# Selecciona la liga que quieres capturar (cadena exacta como será usada en la ruta)
selected_league = "ARGENTINA_A"
# Lista de URLs para la liga seleccionada
ligue = ligues[selected_league]
```

```python
output_folder = f"../../../img/input/year/2025/{selected_league}"
os.makedirs(output_folder, exist_ok=True)

if os.path.exists(output_folder):
    print("✅ La ruta existe")
else:
    print("❌ La ruta no existe")
```

```python
output_folder
```

```python
headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/114.0.0.0 Safari/537.36",
    "Accept-Language": "es-ES,es;q=0.9,en;q=0.8",
}

for url in ligue:
    response = requests.get(url, headers=headers)
    # Verificar si la solicitud fue exitosa
    if response.status_code != 200:
        print(f"❌ Error al acceder a la página {url}. Código: {response.status_code}")
        continue

    soup = BeautifulSoup(response.text, "html.parser")

    # Buscar cada fila de jugador en la tabla
    rows = soup.select("table.items tbody tr")

    for row in rows:
        img_tag = row.select_one("img.bilderrahmen-fixed")
        name_tag = row.select_one("td.hauptlink a")
        country_tag = row.select_one("td.zentriert img.flaggenrahmen")

        if img_tag and name_tag and country_tag:
            img_url = img_tag.get("data-src", img_tag.get("src", "")).split("?")[0]  # Limpiar query
            player_name = name_tag.text.strip()
            country = country_tag.get("alt", "").upper().replace(" ", "_")

            # Procesar nombre: ALAN AGUIRRE → A_AGUIRRE
            name_parts = player_name.strip().upper().split()
            if len(name_parts) >= 2:
                name_formatted = f"{name_parts[0][0]}_{name_parts[-1]}"
            else:
                name_formatted = name_parts[0]

            filename = f"{country}_{name_formatted}.png"
            filepath = os.path.join(output_folder, filename)

            # Descargar la imagen
            try:
                r = requests.get(img_url, headers=headers, timeout=10)
                if r.status_code == 200:
                    with open(filepath, "wb") as f:
                        f.write(r.content)
                    print(f"Guardado: {filename}")
                else:
                    print(f"No se pudo descargar: {img_url} (status {r.status_code})")
            except Exception as e:
                print(f"Error descargando {img_url}: {e}")
```
